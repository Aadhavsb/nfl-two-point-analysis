# NFL two-point decision assistant

This notebook evaluates the decision **at the moment after a touchdown**. It does not regrade historical coaching decisions after final outcomes.

```python
from nfl_two_point.cohort import build_conversion_cohort
from nfl_two_point.config import load_config
from nfl_two_point.ingest import load_play_by_play, write_manifest
from nfl_two_point.validation import temporal_split
from nfl_two_point.win_probability import fit_win_probability, game_states, metrics

config = load_config("config/analysis.yaml")
pbp = load_play_by_play(config)
write_manifest(pbp, config, "data/manifests/pbp.json")
frame = pbp.to_pandas()
cohort = build_conversion_cohort(frame)
states = game_states(frame)
splits = temporal_split(states, config.development_end, config.validation_end)
model = fit_win_probability(splits["development"])
metrics(model, splits["test"])
```
